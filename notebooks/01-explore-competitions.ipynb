```python
def get_shots(match_id):
    url = f"https://raw.githubusercontent.com/statsbomb/open-data/master/data/events/{match_id}.json"
    events = pd.read_json(url)
    events["type_name"] = pd.json_normalize(events["type"])["name"]
    shots = events[events["type_name"] == "Shot"].copy()
    shots["x"] = shots["location"].str[0]
    shots["y"] = shots["location"].str[1]
    shots["distance"] = np.sqrt((120 - shots["x"])**2 + (40 - shots["y"])**2)

    shot_details = pd.json_normalize(shots["shot"])
    shots["xg_statsbomb"] = shot_details["statsbomb_xg"]
    shots["outcome"] = shot_details["outcome.name"]

    shots["match_id"] = match_id
    return shots
```

```python
def preparer(shots):
    shots = shots[shots["period"] != 5].reset_index(drop=True)

    details = pd.json_normalize(shots["shot"])
    shots["shot_type"] = details["type.name"]
    shots["body_part"] = details["body_part.name"]
    shots["technique"] = details["technique.name"]

    shots = shots[shots["shot_type"] != "Penalty"].reset_index(drop=True)

    shots["x"] = shots["location"].str[0]
    shots["y"] = shots["location"].str[1]
    shots["distance"] = np.sqrt((120 - shots["x"])**2 + (40 - shots["y"])**2)

    ux = 120 - shots["x"]
    uy = 36 - shots["y"]
    vx = 120 - shots["x"]
    vy = 44 - shots["y"]
    cos_angle = (ux * vx + uy * vy) / (np.sqrt(ux**2 + uy**2) * np.sqrt(vx**2 + vy**2))
    shots["angle"] = np.degrees(np.arccos(cos_angle))

    shots["under_pressure"] = shots["under_pressure"].fillna(0).astype(int)
    shots["is_goal"] = (shots["outcome"] == "Goal").astype(int)

    return shots


data = preparer(shots_big)

print("Tirs conservés :", data.shape)
print("Taux de buts   :", data["is_goal"].mean())
print(data["angle"].isna().sum(), "angles manquants")
```

```python
def mettre_en_premier(serie, reference):
    autres = sorted(v for v in serie.dropna().unique() if v != reference)
    return pd.Categorical(serie, categories=[reference] + autres)


data["shot_type"] = mettre_en_premier(data["shot_type"], "Open Play")
data["body_part"] = mettre_en_premier(data["body_part"], "Right Foot")
data["technique"] = mettre_en_premier(data["technique"], "Normal")

features_big = pd.get_dummies(
    data[["distance", "angle", "under_pressure", "body_part", "shot_type", "technique"]],
    columns=["body_part", "shot_type", "technique"],
    drop_first=True
)

X = features_big
y = data["is_goal"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print(X_train.shape, X_test.shape)
print(list(features_big.columns))
```

```python
model_lr = LogisticRegression(max_iter=2000)
model_lr.fit(X_train, y_train)

model_gb = HistGradientBoostingClassifier(
    max_iter=500,
    learning_rate=0.05,
    max_leaf_nodes=15,
    min_samples_leaf=40,
    l2_regularization=1.0,
    early_stopping=True,
    validation_fraction=0.15,
    n_iter_no_change=25,
    random_state=42
)
model_gb.fit(X_train, y_train)

comparaison = pd.DataFrame({
    "is_goal": y_test,
    "xg_lr": model_lr.predict_proba(X_test)[:, 1],
    "xg_gb": model_gb.predict_proba(X_test)[:, 1],
    "xg_sb": data.loc[X_test.index, "xg_statsbomb"],
})

print("Entraînement terminé")
print("xG StatsBomb manquants :", comparaison["xg_sb"].isna().sum())
print("xG StatsBomb à 0 ou 1 :", ((comparaison["xg_sb"] <= 0) | (comparaison["xg_sb"] >= 1)).sum())
```

```python
def cote(P, U, V):
    return (V[0] - U[0]) * (P[1] - U[1]) - (V[1] - U[1]) * (P[0] - U[0])


def dans_triangle(P, A, B, C):
    d1 = cote(P, A, B)
    d2 = cote(P, B, C)
    d3 = cote(P, C, A)
    negatif = (d1 < 0) or (d2 < 0) or (d3 < 0)
    positif = (d1 > 0) or (d2 > 0) or (d3 > 0)
    return not (negatif and positif)


def compter_defenseurs(ligne):
    frame = ligne["shot"].get("freeze_frame")
    if not frame:
        return np.nan

    A = (ligne["x"], ligne["y"])
    B = (120, 36)
    C = (120, 44)

    n = 0
    for joueur in frame:
        if joueur["teammate"]:
            continue
        if dans_triangle(joueur["location"], A, B, C):
            n += 1
    return n


data["defenseurs_cone"] = data.apply(compter_defenseurs, axis=1)

print(data["defenseurs_cone"].isna().sum(), "tirs sans freeze_frame")
print(data["defenseurs_cone"].value_counts().sort_index())
```

```python
def extraire_freeze(ligne):
    frame = ligne["shot"].get("freeze_frame") or []
    A = (ligne["x"], ligne["y"])
    B = (120, 36)
    C = (120, 44)

    defenseurs = 0
    gardien_dans_cone = 0
    gardien_x, gardien_y = np.nan, np.nan
    distance_min = np.nan

    for joueur in frame:
        if joueur["teammate"]:
            continue

        P = joueur["location"]
        dedans = dans_triangle(P, A, B, C)
        est_gardien = joueur["position"]["name"] == "Goalkeeper"

        if est_gardien:
            gardien_x, gardien_y = P
            gardien_dans_cone = int(dedans)
        elif dedans:
            defenseurs += 1

        d = np.hypot(P[0] - A[0], P[1] - A[1])
        if np.isnan(distance_min) or d < distance_min:
            distance_min = d

    return pd.Series({
        "defenseurs_cone": defenseurs,
        "gardien_dans_cone": gardien_dans_cone,
        "gardien_dist_but": np.hypot(120 - gardien_x, 40 - gardien_y),
        "adversaire_proche": distance_min,
    })


nouvelles = data.apply(extraire_freeze, axis=1)

for colonne in nouvelles.columns:
    data[colonne] = nouvelles[colonne]

print(data[["defenseurs_cone", "gardien_dans_cone", "gardien_dist_but", "adversaire_proche"]].describe())
print()
print("Valeurs manquantes :")
print(data[["gardien_dist_but", "adversaire_proche"]].isna().sum())
```

```python
from sklearn.model_selection import cross_val_score

def score_cv(modele):
    scores = cross_val_score(modele, X3_train, y3_train, cv=5, scoring="neg_log_loss")
    return -scores.mean(), scores.std()

lr_moy, lr_ec = score_cv(LogisticRegression(max_iter=3000))
gb_moy, gb_ec = score_cv(HistGradientBoostingClassifier(
    max_iter=500,
    learning_rate=0.05,
    max_leaf_nodes=15,
    min_samples_leaf=40,
    l2_regularization=1.0,
    early_stopping=True,
    validation_fraction=0.15,
    n_iter_no_change=25,
    random_state=42
))

print(f"Régression logistique : {lr_moy:.5f}  (écart type {lr_ec:.5f})")
print(f"Gradient boosting     : {gb_moy:.5f}  (écart type {gb_ec:.5f})")
```

```python
def preparer(shots, retirer_penalties=True):
    shots = shots[shots["period"] != 5].reset_index(drop=True)

    details = pd.json_normalize(shots["shot"])
    shots["shot_type"] = details["type.name"]
    shots["body_part"] = details["body_part.name"]
    shots["technique"] = details["technique.name"]

    shots["is_penalty"] = (shots["shot_type"] == "Penalty").astype(int)
    if retirer_penalties:
        shots = shots[shots["is_penalty"] == 0].reset_index(drop=True)

    shots["x"] = shots["location"].str[0]
    shots["y"] = shots["location"].str[1]
    shots["distance"] = np.sqrt((120 - shots["x"])**2 + (40 - shots["y"])**2)

    ux = 120 - shots["x"]
    uy = 36 - shots["y"]
    vx = 120 - shots["x"]
    vy = 44 - shots["y"]
    cos_angle = (ux * vx + uy * vy) / (np.sqrt(ux**2 + uy**2) * np.sqrt(vx**2 + vy**2))
    shots["angle"] = np.degrees(np.arccos(cos_angle))

    shots["under_pressure"] = shots["under_pressure"].fillna(0).astype(int)
    shots["is_goal"] = (shots["outcome"] == "Goal").astype(int)

    shots["team_name"] = pd.json_normalize(shots["team"])["name"]
    shots["player_name"] = pd.json_normalize(shots["player"])["name"]

    nouvelles = shots.apply(extraire_freeze, axis=1)
    for colonne in nouvelles.columns:
        shots[colonne] = nouvelles[colonne]

    return shots


mediane_gardien = data["gardien_dist_but"].median()

affichage = preparer(shots_big, retirer_penalties=False)
affichage["gardien_dist_but"] = affichage["gardien_dist_but"].fillna(mediane_gardien)

print("Tirs pour l'affichage :", len(affichage))
print("dont penalties :", affichage["is_penalty"].sum())
```
