```python
match_ids = matches["match_id"].tolist()
len(match_ids)
```

```python
def get_shots(match_id):
    url = f"https://raw.githubusercontent.com/statsbomb/open-data/master/data/events/{match_id}.json"
    events = pd.read_json(url)
    events["type_name"] = pd.json_normalize(events["type"])["name"]
    shots = events[events["type_name"] == "Shot"].copy()
    shots["x"] = shots["location"].str[0]
    shots["y"] = shots["location"].str[1]
    shots["distance"] = np.sqrt((120 - shots["x"])**2 + (40 - shots["y"])**2)

    shot_details = pd.json_normalize(shots["shot"])
    shots["xg_statsbomb"] = shot_details["statsbomb_xg"]
    shots["outcome"] = shot_details["outcome.name"]

    shots["match_id"] = match_id
    return shots
```

```python
test = get_shots(3857276)
test[["match_id", "distance", "xg_statsbomb", "outcome"]]
```

```python
all_shots = []

for match_id in match_ids:
    print(match_id)
    all_shots.append(get_shots(match_id))

shots_all = pd.concat(all_shots, ignore_index=True)
shots_all.shape
```

```python
shots_all["is_goal"] = (shots_all["outcome"] == "Goal").astype(int)
shots_all["is_goal"].mean()
```

```python
shots_all["is_goal"].sum()
```

```python
shots_all["period"].value_counts()
```

```python
shots_open = shots_all[shots_all["period"] != 5].copy()

print(shots_open.shape)
print(shots_open["is_goal"].sum())
```

```python
shots_open["zone"] = pd.cut(
    shots_open["distance"],
    bins=[0, 6, 12, 18, 24, 30, 100],
    labels=["0-6", "6-12", "12-18", "18-24", "24-30", "30+"]
)

shots_open.groupby("zone", observed=True)["is_goal"].agg(["count", "mean"])
```

```python
ux = 120 - shots_open["x"]
uy = 36 - shots_open["y"]

vx = 120 - shots_open["x"]
vy = 44 - shots_open["y"]

produit_scalaire = ux * vx + uy * vy
norme_u = np.sqrt(ux**2 + uy**2)
norme_v = np.sqrt(vx**2 + vy**2)

cos_angle = produit_scalaire / (norme_u * norme_v)
shots_open["angle"] = np.degrees(np.arccos(cos_angle))
```

```python
shots_open[["x", "y", "distance", "angle", "xg_statsbomb"]].sort_values("angle", ascending=False).head(10)
```

```python
shots_open = shots_open.reset_index(drop=True)

details = pd.json_normalize(shots_open["shot"])

shots_open["shot_type"] = details["type.name"]
shots_open["body_part"] = details["body_part.name"]
shots_open["technique"] = details["technique.name"]

shots_open[["shot_type", "body_part", "technique"]].head()
```

```python
shots_open.groupby("shot_type", observed=True)["is_goal"].agg(["count", "mean"])
```

```python
shots_model = shots_open[shots_open["shot_type"] != "Penalty"].copy()
shots_model.shape
```

```python
shots_model["under_pressure"].value_counts(dropna=False)
```

```python
shots_model["under_pressure"] = shots_model["under_pressure"].fillna(0).astype(int)

features = pd.get_dummies(
    shots_model[["distance", "angle", "under_pressure", "body_part", "shot_type", "technique"]],
    columns=["body_part", "shot_type", "technique"],
    drop_first=True
)

print(features.shape)
features.head()
```
