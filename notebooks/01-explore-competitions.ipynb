```python
events_url = "https://raw.githubusercontent.com/statsbomb/open-data/master/data/events/3857276.json"
events = pd.read_json(events_url)
events.shape
```

```python
events["type_name"] = pd.json_normalize(events["type"])["name"]
events["type_name"].value_counts()
```

```python
shots = events[events["type_name"] == "Shot"]
shots.shape
```

```python
shots["shot"].iloc[0]
```

```python
shots_clean = pd.json_normalize(shots["shot"])
shots_clean[["statsbomb_xg", "body_part.name", "type.name", "outcome.name"]]
```
